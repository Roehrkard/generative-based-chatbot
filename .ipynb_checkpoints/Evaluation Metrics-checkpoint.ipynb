```python
!pip install transformers sentence_transformers ipywidgets flask datasets
!jupyter nbextension enable --py widgetsnbextension

import sys
import json
import logging
import torch
import ipywidgets as widgets

from IPython.display import display
from transformers import BertTokenizer, BertForQuestionAnswering
from sentence_transformers import SentenceTransformer, util
from IPython.display import display, clear_output

# Set up logging
logging.basicConfig(level=logging.INFO)

# Enable GPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# Download the SQuAD v2.0 dataset
# !wget -q https://rajpurkar.github.io/SQuAD-explorer/dataset/train-v2.0.json

# Load the SQuAD v2.0 dataset
with open('train-v2.0.json', 'r') as f:
    data = json.load(f)

# Extract contexts from the dataset
contexts = []
for article in data['data']:
    for paragraph in article['paragraphs']:
        contexts.append(paragraph['context'])

# Load pre-trained BERT model and tokenizer for QA
tokenizer = BertTokenizer.from_pretrained('bert-large-uncased-whole-word-masking-finetuned-squad')
model = BertForQuestionAnswering.from_pretrained('bert-large-uncased-whole-word-masking-finetuned-squad')
model.to(device)  # Move model to GPU

# Load SBERT model for improved context matching
sbert_model = SentenceTransformer('all-MiniLM-L6-v2', device=device)

# Encode all contexts using SBERT for fast similarity search
context_embeddings = sbert_model.encode(contexts, convert_to_tensor=True, device=device, show_progress_bar=True)

def find_best_context(question):
    """Find the most relevant context using SBERT-based cosine similarity."""
    question_embedding = sbert_model.encode(question, convert_to_tensor=True, device=device)
    similarities = util.pytorch_cos_sim(question_embedding, context_embeddings).flatten()
    best_index = similarities.argmax().item()
    return contexts[best_index], similarities[best_index].item()  # Return context and its similarity score

def generate_answer(question, context):
    """Generate an answer using the BERT QA model."""
    # Tokenize with truncation to handle long sequences
    inputs = tokenizer(
        question,
        context,
        return_tensors='pt',
        max_length=512, # I added this due to BERT's limitation
        truncation=True  # Enable truncation
    ).to(device)
    input_ids = inputs['input_ids'].tolist()[0]

    with torch.no_grad():
        outputs = model(**inputs)

    answer_start_scores = outputs.start_logits
    answer_end_scores = outputs.end_logits

    # Find the most likely start and end of the answer span
    answer_start = torch.argmax(answer_start_scores)
    answer_end = torch.argmax(answer_end_scores) + 1

    # Confidence threshold
    confidence_threshold = 0.4

    # Check if the answer is valid
    start_score = answer_start_scores[0, answer_start].item()
    if start_score < confidence_threshold:
        return "I'm not confident about the answer. Can you please ask differently?"

    # Ensure that the end index is greater than the start index
    if answer_end <= answer_start:
        return "I'm sorry, but I couldn't find an answer in the context provided."

    answer = tokenizer.convert_tokens_to_string(
        tokenizer.convert_ids_to_tokens(input_ids[answer_start:answer_end])
    )

    # Check for coherence of the answer
    if not answer.strip():  # If the answer is empty
        return "I'm sorry, but I couldn't find an answer in the context provided."

    return answer

# Create a simple chat interface using ipywidgets
def chat_with_bot():
    output = widgets.Output()
    question_input = widgets.Text(
        description='You:',
        placeholder='Type your question here...'
    )

    display(question_input, output)

    def on_submit(change):
        user_input = change['new']
        if not user_input.strip():
            return

        # Find the most relevant context and its similarity score
        context, similarity_score = find_best_context(user_input)
        logging.info(f"User question: {user_input} | Best context similarity: {similarity_score:.2f}")

        # Generate the answer
        answer = generate_answer(user_input, context)

        # Clear the output and display the interaction
        with output:
            clear_output(wait=True)
            print(f"You: {user_input}")
            print(f"Bot: {answer}")

        # Clear the input for the next question
        question_input.value = ''

    question_input.observe(on_submit, names='value')

# Start the chat interface
chat_with_bot()
```

Evaluation code

```python
from collections import Counter
import re
import string

# Download SQuAD v2.0 dev dataset
!wget -q https://rajpurkar.github.io/SQuAD-explorer/dataset/dev-v2.0.json

# Load development dataset from JSON file
with open('dev-v2.0.json', 'r') as f:
    dev_data = json.load(f)

# Initialize lists to store contexts, questions, and answers from dataset
dev_contexts = []  # Paragraph contexts where questions are asked
dev_questions = []  # Questions based on the contexts
dev_answers = []  # Ground truth answers to the questions (empty if no answer)

# Extract data from each article in dataset
for article in dev_data['data']:
    for paragraph in article['paragraphs']:
        context = paragraph['context']  # Get the paragraph context
        for qa in paragraph['qas']:
            question = qa['question']  # Get the question for this context
            dev_questions.append(question)
            dev_contexts.append(context)

            # Check if the question is unanswerable (SQuAD v2.0 includes these)
            if qa['is_impossible']:
                dev_answers.append([])  # Append an empty list for unanswerable questions
            else:
                # Append the list of valid answers for answerable questions
                answers = [answer['text'] for answer in qa['answers']]
                dev_answers.append(answers)

def normalize_answer(s):
    """
    Normalize a given text by applying these transformations:
    1. Lowercase the text.
    2. Remove punctuation characters.
    3. Remove articles ('a', 'an', 'the').
    4. Remove extra whitespaces.

    Args:
        s (str): The input string to normalize.

    Returns:
        str: The normalized string.
    """
    def remove_articles(text):
        return re.sub(r'\b(a|an|the)\b', ' ', text)

    def white_space_fix(text):
        return ' '.join(text.split())

    def remove_punc(text):
        exclude = set(string.punctuation)
        return ''.join(ch for ch in text if ch not in exclude)

    def lower(text):
        return text.lower()

    return white_space_fix(remove_articles(remove_punc(lower(s))))

def compute_exact(a_gold, a_pred):
    """
    Compute the Exact Match (EM) score between gold standard answer and predicted answer.

    Exact match occurs if normalized versions of both answers are identical.

    Args:
        a_gold (str): The ground truth answer.
        a_pred (str): The predicted answer.

    Returns:
        int: 1 if the answers match exactly, 0 otherwise.
    """
    return int(normalize_answer(a_gold) == normalize_answer(a_pred))

def compute_f1(a_gold, a_pred):
    """
    Computes F1 score between the gold standard answer & predicted answer.

    F1 score is harmonic mean of precision & recall, where:
    - Precision = (# of overlapping words) / (total words in predicted answer)
    - Recall = (# of overlapping words) / (total words in gold standard answer)

    Args:
        a_gold (str): Ground truth answer.
        a_pred (str): Predicted answer.

    Returns:
        float: F1 score between 0 & 1
    """
    gold_toks = normalize_answer(a_gold).split()
    pred_toks = normalize_answer(a_pred).split()
    common = Counter(gold_toks) & Counter(pred_toks)
    num_same = sum(common.values())

    # Edge case: if both gold & predicted answers are empty, score is 1
    if len(gold_toks) == 0 and len(pred_toks) == 0:
        return 1.0
    # If only one answer is empty, score is 0
    if len(gold_toks) == 0 or len(pred_toks) == 0:
        return 0.0
    # If there's no common tokens, the F1 score is 0
    if num_same == 0:
        return 0.0

    precision = num_same / len(pred_toks)
    recall = num_same / len(gold_toks)
    f1 = (2 * precision * recall) / (precision + recall)
    return f1

def evaluate(dev_contexts, dev_questions, dev_answers, num_samples=1000):
    """
    Evaluate the performance of a question-answering model on the SQuAD v2.0 development set.

    The evaluation calculates the Exact Match (EM) & F1 scores for model predictions.

    Args:
        dev_contexts (list of str): List of contexts (paragraphs) from the dataset.
        dev_questions (list of str): List of questions based on the contexts.
        dev_answers (list of list of str): List of ground truth answers, where each entry is
                                           a list of valid answers (or empty for unanswerable questions).
        num_samples (int): The number of samples to evaluate (for speed). Default is 1000.

    Prints:
        Exact Match (EM) score & F1 score as percentages.
    """
    total_em = 0  # Sum of exact matches
    total_f1 = 0  # Sum of F1 scores
    total = 0  # Total number of evaluated samples

    for i in range(len(dev_questions)):
        question = dev_questions[i]
        context = dev_contexts[i]
        ground_truths = dev_answers[i]

        if not ground_truths:
            # Handle unanswerable questions (SQuAD v2.0)
            predicted_answer = generate_answer(question, context)
            if predicted_answer.lower().strip() in ["i'm not confident about the answer. can you please ask differently?",
                                                    "i'm sorry, but i couldn't find an answer in the context provided."]:
                em = 1  # Treat model's inability to answer as correct
                f1 = 1
            else:
                em = 0
                f1 = 0
        else:
            # Evaluate answerable questions
            predicted_answer = generate_answer(question, context)
            em = max(compute_exact(a, predicted_answer) for a in ground_truths)
            f1 = max(compute_f1(a, predicted_answer) for a in ground_truths)

        total_em += em
        total_f1 += f1
        total += 1

        #  Prints results at intervals (can stop running when changes become minimal)
        if (i + 1) % 200 == 0:
            print(f"Processed {i + 1}/{len(dev_questions)} samples")

        # Limits evaluation to a specified number of samples for speed
        if i + 1 >= num_samples:
            break

    # Calculate and print final evaluation scores
    em_score = (total_em / total) * 100
    f1_score = (total_f1 / total) * 100
    print(f"Exact Match (EM): {em_score:.2f}%")
    print(f"F1 Score: {f1_score:.2f}%")

# Call the evaluation function to evaluate the model on the dev set
evaluate(dev_contexts, dev_questions, dev_answers, num_samples=5000)
```
