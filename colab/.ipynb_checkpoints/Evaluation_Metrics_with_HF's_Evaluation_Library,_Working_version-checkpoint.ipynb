```python
# Install necessary packages
!pip install transformers sentence_transformers ipywidgets flask datasets

# Enable ipywidgets
import sys
!jupyter nbextension enable --py widgetsnbextension

# Import libraries
import json
import logging
import torch
from transformers import BertTokenizer, BertForQuestionAnswering
from sentence_transformers import SentenceTransformer, util
import ipywidgets as widgets
from IPython.display import display, clear_output

# Set up logging
logging.basicConfig(level=logging.INFO)

# Enable GPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# Download the SQuAD v2.0 dev dataset
!wget -q https://rajpurkar.github.io/SQuAD-explorer/dataset/dev-v2.0.json

# Load the SQuAD v2.0 dev dataset
with open('dev-v2.0.json', 'r') as f:
    data = json.load(f)

# Extract data from the dataset
qa_data = []

for article in data['data']:
    for paragraph in article['paragraphs']:
        context = paragraph['context']
        for qa in paragraph['qas']:
            question = qa['question']
            id_ = qa['id']
            answers = qa['answers']
            is_impossible = qa.get('is_impossible', False)
            qa_data.append({
                'id': id_,
                'question': question,
                'context': context,
                'answers': answers,
                'is_impossible': is_impossible
            })

# Optionally, limit the number of examples if computational resources are limited
# qa_data = qa_data[:1000]

# Load pre-trained BERT model and tokenizer for QA
tokenizer = BertTokenizer.from_pretrained('bert-large-uncased-whole-word-masking-finetuned-squad')
model = BertForQuestionAnswering.from_pretrained('bert-large-uncased-whole-word-masking-finetuned-squad')
model.to(device)  # Move model to GPU

# Load SBERT model for improved context matching
sbert_model = SentenceTransformer('all-MiniLM-L6-v2', device=device)

# Encode all contexts using SBERT for fast similarity search
contexts = [sample['context'] for sample in qa_data]
context_embeddings = sbert_model.encode(contexts, convert_to_tensor=True, device=device, show_progress_bar=True)

def find_best_context(question):
    """Find the most relevant context using SBERT-based cosine similarity."""
    question_embedding = sbert_model.encode(question, convert_to_tensor=True, device=device)
    similarities = util.pytorch_cos_sim(question_embedding, context_embeddings).flatten()
    best_index = similarities.argmax().item()
    return contexts[best_index], similarities[best_index].item()  # Return context and its similarity score

def generate_answer(question, context):
    """Generate an answer using the BERT QA model."""
    # Tokenize with truncation to handle long sequences
    inputs = tokenizer(
        question,
        context,
        return_tensors='pt',
        max_length=512,  # Due to BERT's limitation
        truncation=True  # Enable truncation
    ).to(device)
    input_ids = inputs['input_ids'].tolist()[0]

    with torch.no_grad():
        outputs = model(**inputs)

    answer_start_scores = outputs.start_logits
    answer_end_scores = outputs.end_logits

    # Find the most likely start and end of the answer span
    answer_start = torch.argmax(answer_start_scores)
    answer_end = torch.argmax(answer_end_scores) + 1

    # Confidence threshold
    confidence_threshold = 0.0  # Set to 0.0 to include all answers

    # Check if the answer is valid
    start_score = answer_start_scores[0, answer_start].item()
    if start_score < confidence_threshold:
        return ""

    # Ensure that the end index is greater than the start index
    if answer_end <= answer_start:
        return ""

    answer = tokenizer.convert_tokens_to_string(
        tokenizer.convert_ids_to_tokens(input_ids[answer_start:answer_end])
    )

    # Check for coherence of the answer
    if not answer.strip():  # If the answer is empty
        return ""

    return answer.strip()

# Generate predictions
from tqdm.notebook import tqdm

predictions = {}
for idx, sample in enumerate(tqdm(qa_data)):
    id_ = sample['id']
    question = sample['question']
    # Use the context provided in the sample
    context = sample['context']
    # Generate the answer
    answer = generate_answer(question, context)
    predictions[id_] = answer

# Save the predictions to a JSON file
with open('predictions.json', 'w') as f:
    json.dump(predictions, f)

# Download the official SQuAD Evaluation Script
!wget -q https://raw.githubusercontent.com/rajpurkar/SQuAD-explorer/master/evaluate-v2.0.py

# Run the evaluation
!python evaluate-v2.0.py dev-v2.0.json predictions.json
```

```python
# Install necessary packages
!pip install transformers sentence_transformers ipywidgets flask datasets evaluate

# Enable ipywidgets
import sys
!jupyter nbextension enable --py widgetsnbextension

# Import libraries
import json
import logging
import torch
from transformers import BertTokenizer, BertForQuestionAnswering
from sentence_transformers import SentenceTransformer, util
import ipywidgets as widgets
from IPython.display import display, clear_output
import evaluate
import torch.nn.functional as F

# Set up logging
logging.basicConfig(level=logging.INFO)

# Enable GPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# Download the SQuAD v2.0 dataset
!wget -q https://rajpurkar.github.io/SQuAD-explorer/dataset/train-v2.0.json

# Load the SQuAD v2.0 dataset
with open('train-v2.0.json', 'r') as f:
    data = json.load(f)

# Extract data from the dataset
qa_data = []

for article in data['data']:
    for paragraph in article['paragraphs']:
        context = paragraph['context']
        for qa in paragraph['qas']:
            question = qa['question']
            id_ = qa['id']
            answers = qa['answers']
            is_impossible = qa.get('is_impossible', False)
            qa_data.append({
                'id': id_,
                'question': question,
                'context': context,
                'answers': answers,
                'is_impossible': is_impossible
            })

# For testing purposes, limit to the first 1000 examples
qa_data = qa_data[:1000]

# Load pre-trained BERT model and tokenizer for QA (fine-tuned on SQuAD v2.0)
tokenizer = BertTokenizer.from_pretrained('deepset/bert-large-uncased-whole-word-masking-squad2')
model = BertForQuestionAnswering.from_pretrained('deepset/bert-large-uncased-whole-word-masking-squad2')
model.to(device)  # Move model to GPU

# Load SBERT model for improved context matching (optional for evaluation)
# sbert_model = SentenceTransformer('all-MiniLM-L6-v2', device=device)
# contexts = [sample['context'] for sample in qa_data]
# context_embeddings = sbert_model.encode(contexts, convert_to_tensor=True, device=device, show_progress_bar=True)

def generate_answer(question, context):
    """Generate an answer and compute no_answer_probability using the BERT QA model."""
    # Tokenize inputs with truncation
    inputs = tokenizer(
        question,
        context,
        return_tensors='pt',
        max_length=512,
        truncation=True
    ).to(device)
    input_ids = inputs['input_ids']
    input_ids_list = input_ids[0].tolist()

    with torch.no_grad():
        outputs = model(**inputs)

    # Get logits
    start_logits = outputs.start_logits[0]
    end_logits = outputs.end_logits[0]

    # Compute the score for the [CLS] token
    cls_index = input_ids_list.index(tokenizer.cls_token_id)
    no_answer_score = (start_logits[cls_index] + end_logits[cls_index]).item()

    # Find the best non-null answer span
    max_answer_length = 30
    n_best_size = 20

    # Convert logits to probabilities
    start_logits = start_logits.cpu().numpy()
    end_logits = end_logits.cpu().numpy()

    # Get the top start and end indices
    start_indices = (-start_logits).argsort()[:n_best_size].tolist()
    end_indices = (-end_logits).argsort()[:n_best_size].tolist()

    valid_answers = []

    for start_index in start_indices:
        for end_index in end_indices:
            if start_index >= len(input_ids_list):
                continue
            if end_index >= len(input_ids_list):
                continue
            if end_index < start_index:
                continue
            if end_index - start_index + 1 > max_answer_length:
                continue

            answer_score = start_logits[start_index] + end_logits[end_index]

            tokens = input_ids_list[start_index:end_index + 1]
            answer = tokenizer.decode(tokens, skip_special_tokens=True, clean_up_tokenization_spaces=True)

            valid_answers.append({
                'text': answer,
                'score': answer_score
            })

    if valid_answers:
        best_answer = max(valid_answers, key=lambda x: x['score'])
        best_answer_text = best_answer['text']
        best_answer_score = best_answer['score']
    else:
        best_answer_text = ''
        best_answer_score = float('-inf')

    # Compute the difference between no answer and best answer scores
    score_diff = no_answer_score - best_answer_score

    # Convert score difference to no_answer_probability
    no_answer_probability = 1 / (1 + torch.exp(-torch.tensor(score_diff))).item()

    # Decide whether to return the answer or an empty string
    if no_answer_probability > 0.5:
        # Model predicts no answer
        final_answer = ''
    else:
        # Model predicts an answer
        final_answer = best_answer_text

    return final_answer, no_answer_probability

# Generate predictions
from tqdm.notebook import tqdm

predictions_list = []

for idx, sample in enumerate(tqdm(qa_data)):
    id_ = sample['id']
    question = sample['question']
    context = sample['context']  # Use the context from the dataset
    # Generate the answer and no_answer_probability
    answer, no_answer_probability = generate_answer(question, context)
    predictions_list.append({
        'id': id_,
        'prediction_text': answer,
        'no_answer_probability': no_answer_probability
    })

# Prepare references
references_list = []
for sample in qa_data:
    answers = sample['answers']
    references_list.append({
        'id': sample['id'],
        'answers': {
            'text': [ans['text'] for ans in answers],
            'answer_start': [ans['answer_start'] for ans in answers]
        }
    })

# Load the SQuAD v2.0 metric
metric = evaluate.load('squad_v2')

# Compute the evaluation metrics
results = metric.compute(predictions=predictions_list, references=references_list)
print(results)
```
